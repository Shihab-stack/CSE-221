```python
import random, time, sys

def insertionSort (A):
    n = len(A)
    for i in range(1,n):
        x = A[i]
        j=i-1
        while j>-1 and x < A[j]:
            A[j+1] = A[j]
            j-=1
        A[j+1]= x
    return A

if __name__=="__main__":
    

    num = 20000
    a = [random.randrange(1000000) for _ in range(num)]

    begin = time.time()
    print(a)
    # b = bubbleSort(a)
    # b = selectionSort(a)
    b = insertionSort(a)
    print(b)
    assert b == sorted(a)
    print("Elapsed time for Bubble/Selection/Insertion Sort: %.3f ms" % ((time.time() - begin) * 1000))

                
```

```python

import asyncio

async def fetch_data(delay):
    print("Fetching data...")
    await asyncio.sleep(delay)
    print("Data fetched")
    return {"Data": "Some data"}

async def main():
    print("Start of main coroutine")

    task = fetch_data(2)
    result = await task
    
    print("End of coroutine")
    await task

# Run the main coroutine
await main()

```

```python
#bubble sort 
def bubbleSort(arr):
    n = len(arr)
    while n>1:   #n-1
        swapped = False 
        for i in range(n-1):
            if arr[i]>arr[i+1]:
                arr[i], arr[i+1] = arr[i+1] ,  arr[i]
                swapped= True
        if not swapped:    #for optimization
             break;            
    return arr    

def selectionSort(arr):
    swapcount =0
    n= len(arr)
    for i in range(n-1):
        smallest = i
        for j in range(i+1,n-1):
            if arr[j]<arr[i]:
                smallest=j
        if smallest!=i:
            arr[smallest] , arr[i] = arr[i] , arr[smallest]
            swapcount+=1
    return zip(arr,arr)   

def insertionSort(arr):
    n = len(arr)
    for i in range(1, n):
        X= arr[i]
        j = i-1
        while j >= 0 and arr[j] > X: # can be fast or slow
            arr[j+1] = arr[j] # make a place for X
            j -= 1
        arr[j+1] = X
    return arr



l= [1,5,2,5,2,6,7,9,-1]
print(bubbleSort(l))
s=""
print(insertionSort(l))


```
